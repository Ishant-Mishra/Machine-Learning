```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import csv
from collections import Counter
```

```python
with open('data.csv') as csv_file:
    csv_reader  = csv.DictReader(csv_file) # for  importing csv files

    # row = next(csv_reader)
    # print(row)

    language_counter = Counter()

    for row in csv_reader:
        language_counter.update(row['LanguagesWorkedWith'].split(';'))

# print(language_counter)
print(language_counter.most_common(15)) # will return a list of 15 most common languages
```

### Plotting a graph

```python
languages = []
popularity = []

for item in language_counter.most_common(15):
    languages.append(item[0])
    popularity.append(item[1])
print(languages)
print(popularity)
```

```python
plt.barh(languages[::-1],popularity[::-1])
plt.title("Most Popular languages")
plt.ylabel("Programming languages")
plt.xlabel("No. of people who uses")

plt.show()
```

### Extracting data through csv file

```python
data = pd.read_csv('data.csv')

ids = data['Responder_id']
lang_responses  = data['LanguagesWorkedWith']

language_counter = Counter()

for response in lang_responses:
    language_counter.update(response.split(';'))

#rest all will be same 
languages = []
popularity = []

for item in language_counter.most_common(15):
    languages.append(item[0])
    popularity.append(item[1])
```
