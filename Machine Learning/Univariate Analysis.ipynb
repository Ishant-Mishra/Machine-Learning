## All about Univariate Analysis

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('train.csv')
```

```python
df.sample(5)
```

## 1. Categorical Data

### a. Countplot

```python
sns.countplot(data=df,x='Pclass')
```

```python
sns.countplot(x='Sex', hue='Survived', data=df)
```

```python
sns.countplot(data=df, x='Pclass' , hue='Survived')
```

```python
sns.countplot(data=df, x='Embarked', hue='Embarked')
```

### b. Piechart

```python
df['Sex'].value_counts().plot(kind='pie',  autopct='%.2f')
```

## 2. Numerical Data

### a. Histogram

```python
sns.histplot(data=df, x='Age',bins=20, element='step')
```

```python
import matplotlib.pyplot as plt
plt.hist(data=df, x='Age')
```

```python
sns.displot(data=df, x='Age', kde=True, element='step',  stat='density')
# The curve is known as PDF (Probability Density Function)
# We caan also check skewness of data using this pdf
```

### c. BoxPlot

```python
sns.boxplot(data=df, x='Fare')
```

```python
sns.boxplot(data=df, x='Age')
```
