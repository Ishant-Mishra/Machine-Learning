### Date and time is by default string when importing from pandas dataframe , first we need to covert datatype into date and time 

# Working with Dates

```python
# converting to Datetime datatype
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

## 1.Exact year

```python
date['date_year'] = date['date'].dt.year
date['date_month'] = date['date'].dt.month
date['date_month_name'] = date['date'].dt.month_name()
date['date_day'] = date['date'].dt.day

date.sample(5)
```

```python
#  day of week
date['date_dow'] = date['date'].dt.dayofweek

date['day_name'] = date['date'].dt.day_name()

date.head()
```

```python
# is weekend ??

date['date_is_weekend'] = np.where(date['day_name'].isin(['Sunday', 'Saturday']),1,0)

date.sample(5)
```

## Extract Quarter

```python
date['date_quarter'] = date['date'].dt.quarter
date.sample(5)
```

## Extract Time elapsed between dates

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today-date['date']
```

```python
(today-date['date']).dt.days
```

## Working with Time

```python
# Time is object first we need to convert to datetime
time['date']  = pd.to_datetime(time['date'])
```

```python
time['hour'] = time['date'].dt.hour
time['minute'] = time['date'].dt.minute
time['sec'] = time['date'].dt.second

# Exact time part

time['time'] = time['date'].dt.time

time.head()
```

## Time Difference

```python
today - time['date']
```

```python
# in seconds
(today - time['date'])/np.timedelta64(1,'s')

# in minutes
(today - time['date'])/np.timedelta64(1,'m')

# in hours
(today - time['date'])/np.timedelta64(1,'h')

```
