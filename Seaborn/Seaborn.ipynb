## 1. Relational Plot
<ol>-To see the statistical relation between two or more variables.</ol>
<ol>-Comes under Bivariate Analysis.</ol>

```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
```

### Scatter Plot

```python
# Scatter Plot -> axes level function
# hue lets you split your data into subgroups and color them differently within the same plot.
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='time')
```

```python
# relplot -> figure level function -> square shape
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',size='size')
```

### Line Plot

```python
gap = px.data.gapminder() # Dataset
df = gap[gap['country'] == 'India']
df
```

```python
# axes level function
sns.lineplot(data=df, x ='year', y ='lifeExp')
```

```python
# using relplot
sns.relplot(data=df,x='year', y='lifeExp', kind='line')
```

```python
temp_df = gap[gap['country'].isin(['India','China','Pakistan'])]
temp_df 
```

```python
sns.relplot(data=temp_df, x='year', y='lifeExp', kind='line', hue='country')
```

```python
sns.lineplot(data=temp_df, x='year', y='lifeExp', hue='country')
# check Legend: It is inside the graph
```

```python
# Taking data with different continent
temp_df = gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df
```

```python
sns.relplot(data=temp_df, x='year', y='lifeExp', kind='line', hue='country', style='continent')
```

### Facet Plot

```python
tips
```

```python
#facet plot
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='sex')
# Here col is used like sex ki jitni bhi category hai uske liye alag alag graph bana do
```

```python
# facet plot -> figure level function -> work with relplot
# It will not work with scatterplot or lineplot
sns.relplot(data=tips, x='total_bill',  y='tip', kind='scatter', row='smoker', col='sex')
```

```python
sns.relplot(data=gap, x='lifeExp', y='gdpPercap', col='year', kind='scatter')
```

```python
# col wrap
sns.relplot(data=gap, x='lifeExp', y='gdpPercap', col='year', col_wrap=4 ,kind='scatter')
```

## 2. Distribution Plots

<li>Used for univariate analysis</li>
<li>Used to find out the distribution</li>
<li>Range of the observation</li>
<li>Central Tendency</li>
<li>Is the data bimodal?</li>
<li>Are there outliers?</li>


#### Plots under distribution plot
<li>HistPlot</li>
<li>kdeplot</li>
<li>rugplot</li>

```python
# figure level -> displot
# axis level -> histplot -> kdeplot -> rugplot
```

```python
# plotting univariate histogram
sns.histplot(data=tips, x='total_bill')
```

```python
sns.displot(data=tips, x='total_bill', kind='hist')
```

```python
# bins parameter
sns.displot(data=tips, x='total_bill', kind='hist', bins=15)
```

```python
# It's  also possible to visualize the distribution of a catagorical variable using the logic of a histogram
# Discrete bins are automatically set for categorical variables
sns.displot(data=tips, x='day', kind='hist')
# Its a kind of count plot
```

```python
# hue -> parameter
sns.displot(data=tips, x='tip', kind='hist', hue='sex')
```

```python
# element -> step
sns.displot(data=tips, x='tip', kind='hist', hue='sex', element='step')
```

```python
titanic = sns.load_dataset('titanic') # predefined dataset in sns
```

```python
sns.displot(data=titanic, x='age', hue='sex',element='step')
```

```python
# facet using col and row -. will not work with histplot function
sns.displot(data=titanic, x='age', kind='hist', col='sex')
```
